```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score , mean_squared_error

```

```python
df= load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
df.shape
```

```python
X= df.drop(columns=["target"])
y= df["target"]
```

```python
X_train, X_test, y_train ,y_test = train_test_split(X, y , test_size=0.3, random_state=42)
```

```python
model= DecisionTreeRegressor()
model.fit(X_train, y_train)
```

```python
y_pred_train= model.predict(X_train)
y_pred_test = model.predict(X_test)

print(f"MSE Train: {mean_squared_error(y_train, y_pred_train)}")
print(f"MSE Train: {mean_squared_error(y_test, y_pred_test)}")

print(f"R2 Train: {r2_score(y_train, y_pred_train)}")
print(f"R2 Train: {r2_score(y_test, y_pred_test)}")
```

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(18,10))
plot_tree(
    model,
    feature_names=X.columns,
    filled=True
)

plt.tight_layout()
plt.show(
    
)
```
